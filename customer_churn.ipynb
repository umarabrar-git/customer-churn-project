Collecting And Load Dataset

```python
import pandas as pd 
import numpy as np 
df=pd.read_excel("customer_dataset.xlsx")
print(df)
print("--------------------------------------------------------------------------")
```

Understanding the dataset


```python
print(df.shape) #Used to get the number of rows and colums of dataset
print("--------------------------------------------------------------------------")
```

```python

print(df.columns)#Used to get column names of the dataset
print("--------------------------------------------------------------------------")
```

```python
print(df.info())#Use to get the information about the dataset like number of rows, columns, data types and memory usage
print("--------------------------------------------------------------------------")
```

```python
print(df.describe())#Used to get the statistical summary of the dataset like as mean, median, standard deviation, minimum and maximum values of the numerical columns
print("--------------------------------------------------------------------------")
```

```python

print(df.head())#Used to get the first 5 rows of the dataset
print("--------------------------------------------------------------------------")
print(df.tail())#Used to get the last 5 rows of the dataset
print("--------------------------------------------------------------------------")
```

```python

#Used to get the Unique values of the categorical columns
print("Unique values in 'Contract' column:")
print(df["Contract"].unique()) 
print("--------------------------------------------------------------------------")
print("Unique values in Internet_Service column:")
print(df["Internet_Service"].unique())
print("--------------------------------------------------------------------------")
print("Unique values in Churn column:")
print(df["Churn"].unique())
print("--------------------------------------------------------------------------")
```

```python
#used to get the value counts of the categorical columns
print("Value counts in 'Contract' column:")
print(df["Contract"].value_counts())
print("--------------------------------------------------------------------------")
print("Value counts in Internet_Service column:")
print(df["Internet_Service"].value_counts())
print("--------------------------------------------------------------------------")
print("Value counts in Churn column:")
print(df["Churn"].value_counts())
print("--------------------------------------------------------------------------")
```

Checking Data quality for cleaning dataset

```python
#Used to check the Data Quality of the dataset like missing vlaues, duplicate values and outliers
print("Missing values in the dataset:")
print(df.isnull().sum())
print('Percentage of missing values in the dataset:')
missing_percent = df.isnull().mean() * 100
print(missing_percent)
print("--------------------------------------------------------------------------")
```

```python
# Check for duplicate values in the entire dataset
print("Duplicate values in the dataset:")
print(df.duplicated().sum())
print("--------------------------------------------------------------------------")
```

```python
# Check for duplicate values in the 'Customer_ID' column
print("Duplicate values in 'Customer_ID' column:")
print(df["Customer_ID"].duplicated().sum())
```

```python
#display the duplicated rows in the dataset
print("Print duplicated rows ")
print(df[df.duplicated()])
print('************************************************************************************')
```

```python
# Check for duplicate values in the 'Internet_Service' column
print('Duplicate Internet_Service')
print(df["Internet_Service"].duplicated().sum())
print('**********************************************************')
```

```python
#value counts of Contract and Internet Service columns
print('value count of Contract :',df["Contract"].value_counts())
print('Value count of Internet Service',df["Internet_Service"].value_counts())
print('----------------------------------------------------------------------------')
```

```python
#describe the Age and Support_Calls columns
print(df["Age"].describe(),"  ",df["Support_Calls"].describe())
print('----------------------------------------------------------------------------')
```

```python
# Check for negative values in 'Age' and 'Support_Calls' columns
print('Age ',df[df['Age']<0],'Support Calls',df[df["Support_Calls"] < 0])
```
