```python
#Outlier detection by using boxplots to visualize the distribution and identify any potential outliers in the "Monthly_Charges" column of the dataset
import seaborn as sns
import matplotlib.pyplot as plt
#to check for all columns in the dataset, we can use a loop to create boxplots for each numerical column. This will help us identify any potential outliers in the data.

for col in numeric_cols:
    sns.boxplot(y=df[col])
    plt.title(f"{col} — Outlier Detection")
    plt.show()
```

```python
#using the interquartile range (IQR) method to identify potential outliers in the numerical columns of the dataset. The IQR is a measure of statistical dispersion, and it can be used to detect values that fall outside of a certain range.
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} potential outliers")
```

```python
#using the Z-score method to identify potential outliers in the numerical columns of the dataset. The Z-score measures how many standard deviations a data point is from the mean, and it can be used to detect values that are significantly different from the rest of the data.
from scipy.stats import zscore

z_scores = zscore(df["Monthly_Charges"].dropna())

outliers = df["Monthly_Charges"].dropna()[
    abs(z_scores) > 3
]

outliers
```

```python

# TIME-SERIES EXPLORATION
import pandas as pd
import matplotlib.pyplot as plt

# Check for datetime columns
datetime_cols = df.select_dtypes(include="datetime").columns

print("Datetime Columns:")
print(datetime_cols)

# If datetime columns exist
if len(datetime_cols) > 0:

    for col in datetime_cols:
        print(f"\n--- {col} ---")
        print("Start Date:", df[col].min())
        print("End Date:", df[col].max())

        # Extract time components
        df[f"{col}_Year"] = df[col].dt.year
        df[f"{col}_Month"] = df[col].dt.month
        df[f"{col}_Day"] = df[col].dt.day
        df[f"{col}_DayOfWeek"] = df[col].dt.day_name()

        print("\nYear Distribution:")
        print(df[f"{col}_Year"].value_counts().sort_index())

        print("\nMonth Distribution:")
        print(df[f"{col}_Month"].value_counts().sort_index())

else:
    print("\nNo datetime column found.")
    print("Time-Series Exploration is not applicable to this dataset.")
```

```python
#TARGET VARIABLE ANALYSIS
target = "Churn"

print("=" * 50)
print("TARGET VARIABLE ANALYSIS")
print("=" * 50)
```

```python

#Check Unique Target Values
print("\n1. Unique Target Values:")
print(df[target].unique())

```

```python

#Target Frequency
print("\n2. Target Frequency:")
target_counts = df[target].value_counts(dropna=False)
print(target_counts)

```

```python

# Target Distribution Visualization

plt.figure(figsize=(6, 4))

df[target].value_counts(dropna=False).plot(kind="bar")

plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.title("Churn Distribution")
plt.xticks(rotation=0)

plt.show()

```

```python
# Check Numerical Features

numeric_cols = df.select_dtypes(include="number").columns

print("\n4. Numerical Features:")
print(list(numeric_cols))
```

```python

# Numerical Features vs Target
print("\n5. Numerical Features vs Churn:")

for col in numeric_cols:

    print(f"\n--- {col} ---")

    print(
        df.groupby(target)[col].agg(
            ["count", "mean", "median", "min", "max"]
        )
    )

    # Box plot
    plt.figure(figsize=(6, 4))

    sns.boxplot(
        x=target,
        y=col,
        data=df
    )

    plt.title(f"{col} vs {target}")
    plt.xlabel("Churn")
    plt.ylabel(col)

    plt.show()

```

```python
#Identify Categorical Features

categorical_cols = df.select_dtypes(include="str").columns

print("\n6. Categorical Features:")
print(list(categorical_cols))
```

```python

#Categorical Features vs Target

print("\n7. Categorical Features vs Churn:")

for col in categorical_cols:

    # Skip Customer_ID because it is an identifier
    if col == "Customer_ID" or col == target:
        continue

    print(f"\n--- {col} ---")

    # Frequency table
    print("\nFrequency:")
    print(
        pd.crosstab(
            df[col],
            df[target],
            dropna=False
        )
    )

    # Percentage within each category
    print("\nPercentage:")
    print(
        (
            pd.crosstab(
                df[col],
                df[target],
                normalize="index",
                dropna=False
            ) * 100
        ).round(2)
    )

    # Bar chart
    pd.crosstab(
        df[col],
        df[target]
    ).plot(kind="bar", figsize=(7, 4))

    plt.title(f"{col} vs {target}")
    plt.xlabel(col)
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=45)

    plt.show()
```
